```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.datasets import load_diabetes


diabetes_data = load_diabetes(as_frame=True, scaled=False)
df = diabetes_data.frame


print("--- ZHTOUMENO 2 ---")
print(f"Διαστάσεις DataFrame (γραμμές, στήλες): {df.shape}")
print("\nΟνόματα 10 χαρακτηριστικών (Features):")
print(list(diabetes_data.feature_names))
print("\nΤύποι δεδομένων στηλών:")
print(df.dtypes)
print("\nΒασική περιγραφή μεταβλητής Target:")
print(df["target"].describe())


print("\n--- ZHTOUMENO 3 ---")
ranges_df = pd.DataFrame(
    {"Min": df.min(), "Max": df.max(), "Range": df.max() - df.min()}
)
print("Εύρος τιμών όλων των μεταβλητών:")
print(ranges_df)


print("\n--- ZHTOUMENO 4 ---")
nan_counts = df.isna().sum()
total_nans = nan_counts.sum()

if total_nans > 0:
    df_clean = df.dropna()
    removed_rows = len(df) - len(df_clean)
    print(
        f"Βρέθηκαν {total_nans} NaN τιμές. Αφαιρέθηκαν {removed_rows} εγγραφές."
    )
else:
    print(
        "Δεν βρέθηκαν NaN τιμές στο dataset. Δεν χρειάστηκε απομάκρυνση εγγραφών."
    )


print("\n--- ZHTOUMENO 5 ---")
plt.figure(figsize=(8, 5))
plt.hist(
    df["target"], bins=30, color="skyblue", edgecolor="black", alpha=0.7
)
plt.title("Distribution of Target Variable (Diabetes Progression)")
plt.xlabel("Target Value (Quantitative measure of disease progression)")
plt.ylabel("Frequency")
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.savefig("target_histogram.png", dpi=300, bbox_inches="tight")
plt.show()
```
